## Stakeholder Brief: Timing Your MassKara Trip and Preparing for Festival Demand

### Context
Every October, thousands of visitors travel from Manila to Bacolod for the MassKara Festival. It can then be intuitive to assume that high Google searches for the festival means flight and hotel prices are about to jump, while local Bacolod hotels, restaurants, and vendors need a reliable way to tell early online curiosity apart from real tourist arrival pressure. To answer this, we tracked daily Google search interest for "MassKara Festival" alongside Manila-to-Bacolod (MNL–BCD) one-way flight fares and Bacolod hotel rates for 36 days leading up to the October 9–13 festival.

### Main Finding
**Flight prices stay calm until the final 7 days before the festival—and then jump sharply—while hotel prices stay steady all month long.**

* **Flights have a sharp 7-day "booking cliff":** More than three weeks before MassKara, the cheapest one-way MNL–BCD ticket held steady around **₱1,915**, and typical fares sat at **₱2,457**. Between 8 and 20 days out, typical fares nudged up only slightly to **₱2,695**. Inside the final 7 days, however, typical fares nearly doubled to **₱4,827**, cheapest available seats jumped **74%** (peaking at **₱9,080**, or nearly **5×** the early minimum), and single-day price jumps reached **+₱4,865 overnight**.
* **Hotels price the festival steadily from the start:** Typical Bacolod hotel rates rose gradually from **₱1,885** (more than 20 days out) to **₱1,960** (8–20 days out) and **₱2,133** in the final week—only **₱248** more per night. Hotels maintained a consistent **5% to 10%** festival markup over off-peak rates throughout the entire tracking period.
* **Early search spikes do not raise prices on their own:** Google search interest climbed steadily as October approached, and in the final week high search interest and surging airfares happened at the same time. Earlier in September, however, days with high search buzz had the exact same typical flight price (**₱2,695**) as quiet days—meaning the calendar countdown, not early online buzz, is what triggers the late airfare surge.

### Why It Matters
* **For Budget-Conscious Travelers:** Waiting until the final week to book a flight costs roughly **₱2,370 extra** on a typical ticket (and much more on peak days), whereas waiting on a hotel adds only a modest nightly difference. Knowing this separates real urgency from false alarms: early news or social media buzz in September is not a reason to panic-book, but crossing into the final 7 days before departure carries a heavy financial penalty.
* **For Local Bacolod Businesses:** Because late-booking visitors spend far more on airfare in the final week, they arrive with tighter budgets for dining, tours, and souvenirs. At the same time, tracking daily Google search spikes too early in September can send false signals—steady calendar milestones are a much more reliable guide for staffing and inventory.

### Recommended Actions
1. **Book MNL–BCD flights at least 14 to 21 days before MassKara (and never wait past Day 7):** Travelers should lock in airfare two to three weeks ahead to secure fares near ₱2,450–₱2,700. If travel plans are still undecided inside the final 7 days, consider alternative routes (such as flying to Iloilo and taking a ferry) rather than paying last-minute airfare spikes.
2. **Reserve Bacolod hotels early—and don't wait for last-minute deals:** Because hotels apply a steady 5%–10% festival markup from early September onward, prices do not drop closer to the event. Booking early secures better room choices without paying any extra premium.
3. **Anchor local business prep to the 7-day countdown and bundle early:** Local hotels, restaurants, and tour operators should use the 14-day and 7-day marks before the festival—not isolated September search spikes—to schedule extra staff and stock up on supplies. Offering room-and-dining or tour packages in early September lets businesses capture visitor spending before late-week airfare surges squeeze travelers' budgets.

### Scope & Data Note
These takeaways are based on publicly listed daily quotes for MNL–BCD flights, Bacolod hotels, and Google search interest collected from September 2 to October 8, 2026 (including 6 observation days in the final 7-day window). They describe observed price patterns rather than airline seat counts or internal hotel occupancy.
