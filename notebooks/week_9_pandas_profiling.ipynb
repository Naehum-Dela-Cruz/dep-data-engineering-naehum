```python
# setup stuff
import pandas as pd
```

```python
# flights profiling

# this loads the jsonl file into a data frame
df_flights = pd.read_json("../data/raw/google_flights.jsonl", lines=True)

# this describes or 'profiles' df_flights 
print("Shape:", df_flights.shape)
print("\n--- Info ---")
df_flights.info()

# prints the first three rows
print("\n--- Head ---")
print(df_flights.head(3))

# iirc this counts the number of missing values, perhaps by column?
# lotsa missing target_date due to rolling collection not having a target date (this is expected)
print("\n--- Missing Values ---")
print(df_flights.isnull().sum())
```
