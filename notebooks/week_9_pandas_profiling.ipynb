```python
# setup stuff
import pandas as pd
```

```python
# flights profiling

# this loads the jsonl file into a data frame
df_flights = pd.read_json("../data/raw/google_flights.jsonl", lines=True)

# this describes or 'profiles' df_flights 
print("Shape:", df_flights.shape)
print("\n--- Info ---")
df_flights.info()

# prints the first three rows
print("\n--- Head ---")
print(df_flights.head(3))

# iirc this counts the number of missing values, perhaps by column?
# lotsa missing target_date due to rolling collection not having a target date (this is expected) (now fixed)
print("\n--- Missing Values ---")
# print(df_flights.isnull().sum())

# first exclude rolling target dates
null_counts = df_flights.isnull().sum()
rolling_count = (df_flights["collection_mode"] == "rolling").sum()
null_counts["target_date"] = null_counts["target_date"] - rolling_count

# print new recalculated values
print(null_counts)
```

```python
# standardize to snake case 
df_flights.columns = [col.strip().lower().replace(" ","_") for col in df_flights.columns]
df_hotels.columns = [col.strip().lower().replace(" ","_") for col in df_hotels.columns]
df_trends.columns = [col.strip().lower().replace(" ","_") for col in df_trends.columns]

# clean text prices to numeric
df_hotels["price_php"] = df_hotels["lowest_rate"].astype(str).replace(r"[^\d.]","",regex=True)
df_hotels["price_php"] = pd.to_numeric(df_hotels["price_php"], errors="coerce")
```

```python
# normalize nested json objects
def extract_flight_leg_info(row):
    flights_data = row.get("flights")
    if isinstance(flights_data, list) and len(flights_data) > 0:
        leg = flights_data[0]
        dep_airport = leg.get("departure_airport") or {}
        arr_airport = leg.get("arrival_airport") or {}
        return pd.Series({
            "airline": leg.get("airline"),
            "flight_number": leg.get("flight_number"),
            "airplane": leg.get("airplane"),
            "travel_class": leg.get("travel_class"),
            "departure_airport": dep_airport.get("id"),
            "departure_time": dep_airport.get("time"),
            "arrival_airport": arr_airport.get("id"),
            "arrival_time": arr_airport.get("time"),
            "leg_duration": leg.get("duration")
        })
    return pd.Series({})

# apply extraction and merge w/df_flights
flight_details = df_flights.apply(extract_flight_leg_info, axis=1)

# drop redundant or raw complex cols
cols_to_drop = list(flight_details.columns) + ["flights", "booking_token", "airline_logo"]

df_flights = df_flights.drop(columns=cols_to_drop, errors="ignore")
df_flights = pd.concat([df_flights, flight_details], axis=1)
df_flights[["airline", "flight_number", "departure_time", "arrival_time","price"]].head()
```

```python
from pathlib import Path

processed_dir = Path("../data/processed")
processed_dir.mkdir(parents=True, exist_ok=True)

df_flights.to_csv(processed_dir / "cleaned_flights.csv", index=False)
df_hotels.to_csv(processed_dir / "cleaned_hotels.csv", index=False)
df_trends.to_csv(processed_dir / "cleaned_trends.csv", index=False)

print("Saved cleaned files to data/processed/")
```
